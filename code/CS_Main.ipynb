**14. Read all json files with the video categories**

```python
import json
from pathlib import Path
import pandas as pd
```

```python
categories = Path("../catagories")
```

```python
json_files = sorted(categories.glob("*_category_id.json"))
print("Founded files:", [p.name for p in json_files])
```

```python
json_files
```

```python
df2 = []
```

```python
for jf in json_files:
    country = jf.stem.split("_")[0].upper()  # CA_category_id -> CA

    with open(jf, "r", encoding="utf-8") as f:
        data = json.load(f)
        
    items = data.get("items", [])
    
    cat_df = pd.json_normalize(items)

    cat_df = cat_df[["id", "snippet.title", "snippet.assignable"]].rename(columns={
        "id": "category_id",
        "snippet.title": "category_name",
        "snippet.assignable": "assignable"
    })

    cat_df["category_id"] = cat_df["category_id"].astype(str)
    cat_df["country"] = country

    df2.append(cat_df)
```

```python
categories_df = pd.concat(df2, ignore_index=True)
categories_df.head()
```

```python
categories_df['country'].unique()
```

We load multiple JSON files containing YouTube video category information, extract the relevant fields from the items section, and normalize them into a tabular format. 
For each file, we have the country code from the filename, keep only the category ID, category name, and assignable flag, and add the corresponding country as a new column. 
Finally, all country-specific datasets are combined into a single DataFrame, which provides a unified view of video categories across all regions.

**15. For each country, determine how many videos have a category that is not assignable.**

```python
df["category_id"] = df["category_id"].astype(str)
categories_df["category_id"] = categories_df["category_id"].astype(str)
```

```python
df_merged = df.merge(
    categories_df[["country", "category_id", "assignable"]],
    on=["country", "category_id"],
    how="left"
)
```

```python
df_merged
```

```python
#Check country alignment
print("df countries:", sorted(df["country"].unique()))
print("cat countries:", sorted(categories_df["country"].unique()))

#Check for duplicates in categories_df
print(categories_df.duplicated(subset=["country", "category_id"]).sum())

#Check merge result size vs original
print("Original df:", df.shape[0])
print("Merged df:", df_merged.shape[0])  # should equal df.shape[0] for a clean left join
```

```python
not_assignable_df = df_merged[df_merged["assignable"] == False]
not_assignable_df.shape
```

```python
videos_not_assigned = (
    not_assignable_df
    .groupby("country")
    .size()
    .reset_index(name="videos_not_assigned")
    .sort_values("videos_not_assigned", ascending=False)
)

videos_not_assigned
```

```python
result = (
    df_merged.assign(not_assignable = df_merged["assignable"] == False)
    .groupby("country")["not_assignable"]
    .sum()
    .reset_index(name="videos_not_assigned")
    .sort_values("videos_not_assigned", ascending=False)
)

result
```

To determine how many videos belong to non-assignable categories per country, we first ensured that the category_id column in both datasets (df and categories_df) had the same data type (string). This was necessary for a correct merge.

Next, we merged the video dataset with the category dataset using country and category_id as keys, adding the assignable information to each video.

After the merge, we filtered the dataset to keep only videos where the category is not assignable (assignable == False), which allowed us to isolate the relevant records.

We then computed the number of such videos per country using two approaches: first by grouping the filtered dataset and counting rows, and second by creating a boolean indicator for non-assignable videos and summing it per country.

Both methods produced the same result, showing the number of videos in non-assignable categories for each country.
